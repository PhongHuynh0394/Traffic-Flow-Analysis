```python
from pyspark.sql import SparkSession

# Connect to Spark master running on docker-net
spark = (
    SparkSession.builder
    .appName("NotebookApp")
    .master("spark://spark-master:7077")  # Connect to the cluster
    .config("spark.executor.memory", "512m")
    .config("spark.executor.cores", "1")
    .getOrCreate()
)

# ✅ Test: show version and run something simple
print("Spark version:", spark.version)
spark.range(10).show()
```

```python
spark.stop()
```
